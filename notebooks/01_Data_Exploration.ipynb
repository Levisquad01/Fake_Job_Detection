```python
import matplotlib.pyplot as plt

missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

plt.figure(figsize=(10,6))
missing.plot(kind="bar")
plt.title("Missing Values per Column")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
counts = df["fraudulent"].value_counts()

plt.figure(figsize=(6,4))
plt.bar(["Real","Fake"], counts)
plt.title("Distribution of Real and Fake Jobs")
plt.ylabel("Number of Jobs")
plt.show()
```

## Initial Observations

- The dataset contains 17,880 job postings and 18 features.
- The target variable is `fraudulent`, indicating whether a job posting is legitimate (0) or fake (1).
- The dataset is highly imbalanced, with approximately 95% legitimate postings and 5% fraudulent postings.
- Several columns contain a significant number of missing values, especially `salary_range` and `department`.
- Most predictive information is expected to come from text columns such as `title`, `description`, `requirements`, and `company_profile`.
- Text preprocessing will be necessary before training machine learning models.
